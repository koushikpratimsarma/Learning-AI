# Natural Language Processing (NLP) and Large Language Models (LLMs)

## Objective
Understand the basics of NLP, the role of Large Language Models, their advantages, limitations, and why language processing is challenging.

---

## What is NLP?

Natural Language Processing (NLP) is a branch of Artificial Intelligence (AI) that combines linguistics and machine learning to help computers understand, interpret, and generate human language.

The goal of NLP is not only to understand individual words but also their meaning and context within a sentence.

---

## Common NLP Tasks

- **Text Classification**
  - Sentiment analysis
  - Spam detection
  - Grammar checking

- **Token (Word) Classification**
  - Part-of-Speech (POS) tagging
  - Named Entity Recognition (NER)

- **Text Generation**
  - Auto-completing text
  - Filling masked words

- **Question Answering**
  - Finding answers from a given document or context

- **Translation & Summarization**
  - Language translation
  - Text summarization

- **Speech & Vision Applications**
  - Speech-to-text transcription
  - Image caption generation

---

## What is a Large Language Model (LLM)?

A Large Language Model (LLM) is an AI model trained on a massive amount of text data to understand and generate human-like language.

Examples:
- GPT
- Llama
- Qwen
- Gemma

LLMs can perform many language tasks without training a separate model for each task.

---

## Key Characteristics of LLMs

- Trained on very large datasets
- Contain millions or billions of parameters
- Perform multiple NLP tasks with a single model
- Learn from examples provided in prompts (In-context Learning)
- Show new abilities as model size increases (Emergent Abilities)

---

## Advantages of LLMs

- One model can solve many NLP tasks.
- Produces natural and fluent text.
- Requires less task-specific training.
- Supports chatbots, translation, summarization, coding, and many other applications.

---

## Limitations of LLMs

- Can generate incorrect information (Hallucination)
- Do not truly understand the world
- May inherit biases from training data
- Limited context window
- Require high computational resources

---

## Why is NLP Difficult?

Human language is complex because it contains:

- Ambiguity
- Context
- Sarcasm
- Humor
- Cultural differences

Humans naturally understand these, but computers need language to be converted into a mathematical representation before they can learn from it.

---

## Key Takeaways

- NLP enables computers to process human language.
- LLMs have significantly improved NLP by using one model for many tasks.
- Despite their capabilities, LLMs still have important limitations.
- Understanding context remains one of the biggest challenges in NLP.
