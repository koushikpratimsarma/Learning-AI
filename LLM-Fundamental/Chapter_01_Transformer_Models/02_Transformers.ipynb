# What is a Transformer?

A **Transformer** is a deep learning architecture introduced in **2017** in the research paper **"Attention Is All You Need"**.

It was designed to solve Natural Language Processing (NLP) tasks more efficiently than previous sequence models like **RNNs** and **LSTMs**.

Today, most modern Large Language Models (LLMs) such as **GPT, BERT, Llama, Gemma, Qwen, and Mistral** are built on the Transformer architecture.

---

# Why Were Transformers Introduced?

Before Transformers, NLP models mainly used:

- Recurrent Neural Networks (RNNs)
- Long Short-Term Memory Networks (LSTMs)

These models had several limitations:

- Processed one word at a time.
- Training was slow.
- Difficult to learn long-range relationships.
- Hard to scale for very large datasets.

Transformers solved these problems by introducing the **Attention Mechanism**, allowing the model to process all words in parallel.

---

| Feature | RNN | LSTM | Transformer |
|---------|-----|------|-------------|

| Parallel Processing | ❌ | ❌ | ✅ |

# Why Are Transformers Powerful?

The biggest innovation is the **Attention Mechanism**.

Instead of reading words one by one, the Transformer looks at the relationship between all words simultaneously.

Example:

**Sentence:**

> The animal didn't cross the road because **it** was tired.

The model understands that **"it"** refers to **"animal"**, not **"road"**.

This ability to understand context makes Transformers much more effective than older NLP models.

# What is a Pipeline?

A **pipeline** is the easiest way to use a pretrained Transformer model.

Instead of writing preprocessing, inference, and post-processing manually, the pipeline handles everything automatically.

### Pipeline Workflow

```text
Input
   │
   ▼
Preprocessing
   │
   ▼
Transformer Model
   │
   ▼
Post-processing
   │
   ▼
Prediction
```

# First Transformer Program

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")

result = classifier("I love learning Transformers!")

print(result)
```

# What Happens Behind the Pipeline?

When we call `pipeline()`, Hugging Face performs three steps automatically.

## 1. Preprocessing

- Takes the input text.
- Converts it into tokens.
- Converts tokens into IDs.

## 2. Model Inference

- Sends token IDs to the pretrained Transformer.
- The model predicts the output.

## 3. Post-processing

- Converts model outputs into human-readable labels and confidence scores.
