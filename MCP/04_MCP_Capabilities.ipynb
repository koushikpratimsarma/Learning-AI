# What are MCP Capabilities?

Capabilities are the features an MCP Server provides to an AI application.

Think of an MCP Server as a smartphone.

A smartphone has different apps:

📞 Phone

📷 Camera

🌐 Browser

🗺 Maps

Each app provides a different capability.

Similarly,

An MCP Server exposes different capabilities to the AI.

## The Four MCP Capabilities

1. Tools
2. Resources
3. Prompts
4. Sampling

# 1. Tools

A Tool is an executable function.

The AI decides when to call it.

Examples

✔ Search Google

✔ Send Email

✔ Query Database

✔ Create GitHub Issue

✔ Get Weather

```python
def get_weather(city):

    return f"The weather in {city} is Sunny."


print(get_weather("Guwahati"))
```

This function performs an action.

Therefore it is a Tool.

---

Imagine asking ChatGPT:

"What is today's weather?"

The LLM cannot know the live weather.

So it calls

Weather Tool

↓

Weather API

↓

Returns Result

# Why are Tools Dangerous?

Tools can change data.

Examples

Delete File

Send Email

Transfer Money

Book Ticket

Because they have side effects,

the Host usually asks for user approval.

```python
def delete_file():

    print("Deleting file...")
```

This changes something.

Therefore it is considered a Tool with side effects.

---

# 2. Resources

Resources provide information.

Unlike Tools,

Resources do not perform actions.

They simply provide read-only data.

```python
notes = {

    "MCP":"Model Context Protocol",

    "LLM":"Large Language Model"

}

print(notes["MCP"])
```

Examples of Resources

📄 PDF

📄 Text File

🗄 Database Record

📚 Documentation

Configuration File

Resources are similar to REST GET endpoints.

---

Difference

Tool--> Reads and changes

Resource--> Only reads

# 3. Prompts

A Prompt is a predefined template.

Instead of writing instructions every time,

the user chooses a ready-made workflow.

```python
def code_review_prompt(code):

    return f"""
You are an expert Python reviewer.

Review the following code.

{code}
"""
```

```python
print(code_review_prompt("print('Hello')"))
```

Examples

Summarize Document

Translate Text

Review Code

Write Email

These are Prompt Templates.

# 4. Sampling

Sampling is the most confusing capability.

Let's understand it with an example.

Normally,

Host

↓

LLM

↓

Answer

But sometimes,

the Server itself wants the LLM to think.

Since the Server cannot directly access the LLM,

it asks the Host.

This process is called Sampling.

---

Server

↓

"Please ask the LLM"

↓

Host

↓

LLM

↓

Answer

↓

Server

```python
class Host:

    def ask_llm(self,prompt):

        print("Host asking LLM...")

        return "This is the AI response."


host = Host()

response = host.ask_llm("Explain MCP")

print(response)
```

Notice

The Server never talks directly to the LLM.

Server

↓

Host

↓

LLM

↓

Host

↓

Server

This is Sampling.

```python
class Server:

    def analyze(self):

        print("Server needs AI help...")

        result = host.ask_llm("Analyze data")

        print(result)


server = Server()

server.analyze()
```

Real Example

Imagine a Resume Analyzer.

Step 1

The Server extracts text.

↓

Step 2

The Server asks

"Host, please let the LLM summarize this resume."

↓

Step 3

Host sends prompt to LLM.

↓

Step 4

LLM returns summary.

↓

Step 5

Host sends summary back to Server.

This is Sampling.

## Capability Comparison

```python
capabilities = {

    "Tool":"Perform Action",

    "Resource":"Read Data",

    "Prompt":"Guide LLM",

    "Sampling":"Ask Host to use LLM"

}

for k,v in capabilities.items():

    print(f"{k:10} -> {v}")
```
