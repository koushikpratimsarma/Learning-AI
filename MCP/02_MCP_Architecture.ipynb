# What is MCP Architecture?

MCP follows a Client-Server Architecture.

There are three main components:

1. Host
2. Client
3. Server

Each component has a different responsibility.

Think of it like a restaurant.

Customer → Waiter → Chef

## Restaurant Analogy

Customer

↓

Waiter

↓

Chef

Customer = User

Restaurant = Host

Waiter = Client

Chef = MCP Server

The waiter never cooks food.

The waiter only carries the request.

Similarly,

The MCP Client never performs the task.

It only communicates with the MCP Server.

## Component 1 : Host

The Host is the application that users interact with.

Examples

- Claude Desktop
- Cursor
- VS Code
- ChatGPT (conceptually)

Responsibilities

- Talks to users
- Uses an LLM
- Connects to MCP Servers
- Shows results

```python
print("Host Started...")

user_question = "What is the weather today?"

print("User:", user_question)
#The Host receives the user's request.
```

## Component 2 : Client

The Client lives inside the Host.

Its only job is communication.

It

✔ connects to MCP Servers

✔ sends requests

✔ receives responses

✔ returns results to the Host

### One Client talks to one MCP Server.

```python
class MCPClient:

    def send_request(self):

        print("Sending request to MCP Server...")
```

```python
client = MCPClient()

client.send_request()
```

The Client does not answer the question.

It only sends the request.

## Component 3 : Server

The MCP Server performs the actual work.

Examples

Weather Server

GitHub Server

Filesystem Server

Database Server

Browser Server

```python
class WeatherServer:

    def get_weather(self):

        return "Sunny, 30°C"
```

```python
server = WeatherServer()

server.get_weather()
```

The Server owns the functionality.

## Communication Flow

```python
class WeatherServer:

    def get_weather(self):

        return "Sunny, 30°C"


class MCPClient:

    def __init__(self, server):

        self.server = server

    def request_weather(self):

        print("Client -> Server")

        return self.server.get_weather()


server = WeatherServer()

client = MCPClient(server)

result = client.request_weather()

print("Server -> Client")

print(result)
```

The Client requests.

The Server responds.

Now let's add the Host.

### Flow

User

↓

Host

↓

Client

↓

Server

↓

Client

↓

Host

↓

User

```python
class WeatherServer:

    def get_weather(self):

        return "Sunny"


class MCPClient:

    def __init__(self, server):

        self.server = server

    def request_weather(self):

        print("Client -> Server")

        return self.server.get_weather()


class Host:

    def __init__(self):

        self.client = MCPClient(WeatherServer())

    def ask(self):

        print("Host received user question")

        answer = self.client.request_weather()

        print("Host shows answer")

        print(answer)


host = Host()

host.ask()
```

Notice something important.

The Host never directly talks to the Server.

The Client is always in the middle.

Host

↓

Client

↓

Server

# Why is this Architecture Powerful?

Suppose we connect multiple servers.

Host

↓

Client 1 → Weather Server

↓

Client 2 → GitHub Server

↓

Client 3 → Filesystem Server

↓

Client 4 → Browser Server

One Host

Many Clients

Many Servers

This makes MCP modular.

```python
servers = [

    "Weather",

    "GitHub",

    "Filesystem",

    "Browser"

]

print("Host connected to")

for server in servers:

    print(server)
```

# Why is it called Modular?

Suppose tomorrow we create a new server.

Calendar Server

Do we need to modify the Host?

No.

Just connect another MCP Server.

Everything continues to work.

This is called modular architecture.

Q2. What is the responsibility of the Host?

Q5. Why is MCP architecture modular?

Because Hosts and Servers are loosely coupled. New Servers can be added without changing existing Hosts.
