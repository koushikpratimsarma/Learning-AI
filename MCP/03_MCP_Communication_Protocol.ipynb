# What is the Communication Protocol?

A communication protocol is a set of rules that two systems follow to exchange information.

Think of it like two people speaking the same language.

If one speaks English and the other speaks Japanese,
they cannot communicate.

Similarly,

The MCP Client and MCP Server communicate using one common language:

**JSON-RPC 2.0**

## Why JSON-RPC?

JSON-RPC is:

✔ Simple

✔ Lightweight

✔ Human readable

✔ Language independent

Every request and response between Client and Server is represented as JSON.

```python
import json

message = {

    "jsonrpc":"2.0",

    "id":1,

    "method":"tools/call"

}

print(json.dumps(message,indent=4))
```

This is the format understood by both the Client and the Server.

# Message Types

The MCP protocol defines three message types.

1. Request

2. Response

3. Notification

## 1. Request

A Request asks the Server to perform some work.

Example

"Please call the weather tool."

```python
request = {

    "jsonrpc":"2.0",

    "id":1,

    "method":"tools/call",

    "params":{

        "tool":"weather"

    }

}

print(json.dumps(request,indent=4))
```

The Client sends this request to the Server.

## 2. Response

The Server processes the request and sends back a response.

The response contains either:

✔ result

or

✔ error

```python
response = {

    "jsonrpc":"2.0",

    "id":1,

    "result":{

        "temperature":"30°C"

    }

}

print(json.dumps(response,indent=4))
```

The Server has completed the request.

## Error Response

If something goes wrong,

the Server returns an error instead of a result.

```python
error = {

    "jsonrpc":"2.0",

    "id":1,

    "error":{

        "message":"Tool not found"

    }

}

print(json.dumps(error,indent=4))
```

## 3. Notification

A notification is a one-way message.

Unlike a request,

it does NOT expect any response.

Example

"Processing..."

"50% completed..."

```python
notification = {

    "jsonrpc":"2.0",

    "method":"progress",

    "params":{

        "progress":"50%"

    }

}

print(json.dumps(notification,indent=4))
```

The Server simply informs the Client.

# Complete Communication Flow

```python
class Server:

    def execute(self):

        print("Server executing tool...")

        return "Sunny"


class Client:

    def __init__(self,server):

        self.server = server

    def request(self):

        print("Request -> Server")

        result = self.server.execute()

        print("Response <- Server")

        return result


client = Client(Server())

print(client.request())
```

This is exactly what happens in MCP.

Request

↓

Server executes

↓

Response

# Interaction Lifecycle

Every MCP connection follows four stages.

1. Initialization

2. Discovery

3. Execution

4. Termination

```python
steps = [

    "Initialize",

    "Discover Capabilities",

    "Execute Tool",

    "Shutdown"

]

for step in steps:

    print(step)
```

## Stage 1 : Initialization

The Client first connects to the Server.

They exchange:

- protocol version

- supported capabilities

Only after successful initialization can communication begin.

```python
print("Client -> initialize")

print("Server -> OK")

print("Connection Established")
```

## Stage 2 : Capability Discovery

The Client asks

"What capabilities do you provide?"

The Server replies

Tools

Resources

Prompts

```python
tools = [

    "Weather",

    "GitHub",

    "Filesystem"

]

print("Available Tools")

for tool in tools:

    print(tool)
```

## Stage 3 : Tool Execution

Now the Client knows the available tools.

It calls one of them.

```python
def weather():

    return "Sunny"

print(weather())
```

## Stage 4 : Termination

After the work is completed,

the Client closes the connection gracefully.

```python
print("Shutdown")

print("Connection Closed")
```
